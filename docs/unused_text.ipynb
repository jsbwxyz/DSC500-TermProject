WigglNet would be an online service that provides mined data about specific product categories to small e-commerce sellers, product manufacturers, and AI trainers. The product (the data) should follow three rules:

 1) Data provided to clients should be in it's most actionable form, in that it should not need to broken down any further to be more useful. For example, the usefulness of any column should be able to be explained by the existance of another column in the table. 
 2) Without altering anything, the data provided to clients should include enough to generate rich visualizations. These visualizations will be generated using matplotlib/seaborn for the email based service, and plotly/ D3.js for the online dashboard service add-on. 
 3) Any data provided to clients should be able to be used in a machine learning model of the client's own. If possible, this version of the data would come at a premimum as there is high demand for already-trained data.

To demonstrate these rule #2, a query will be pulled straight from the database where the pipeline ends. Four meaningful visualizations will be made from this data representing the usefulness.

The pilot product category for WigglNet is Home Decor/Furniture.
